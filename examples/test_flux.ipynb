```python
import torch
from diffusers import FluxPipeline
from numpy.ma.core import negative

device = "cuda:1"
pipe = FluxPipeline.from_pretrained("black-forest-labs/FLUX.1-schnell", torch_dtype=torch.bfloat16).to(device)#save some VRAM by offloading the model to CPU. Remove this if you have enough GPU power


```

```python
prompt = "A single black SOIC-16 integrated circuit, extreme macro close‑up, "
"only the IC body and its 16 gull‑wing pins in frame, blank green FR-4 "
"around it, studio lighting, 8‑k photorealism"
image = pipe(
    prompt,
    negative_prompt="any other components, resistors, capacitors, multiple ICs, text artifacts, watermark, blurry edges, multiple chips, dirt, scratches, low resolution",
    guidance_scale=0.0,
    num_inference_steps=4,
    max_sequence_length=256,
    generator=torch.Generator(device).manual_seed(0)
).images[0]
image
```
